# Exemplo prático de cálculo das métricas de fidelidade

Este exemplo mostra como calcular as métricas POS@K, CDCG@K, INS@K e DEL@K usando funções e dados simples.

```python
# Funções dummy para simular recomendação e score
from metrics.fidelity import pos_at_kr_ke, cdcg_at_ke, ins_at_ke, del_at_ke

def dummy_rank_fn(profile):
    # Retorna itens ordenados por ordem alfabética (simulação)
    return sorted(profile)

def dummy_score_fn(profile, item):
    # Score é o índice do item no perfil (quanto menor, melhor)
    try:
        return len(profile) - profile.index(item)
    except ValueError:
        return 0
```

```python
# Dados de exemplo
user_profile = ['A', 'B', 'C', 'D', 'E'] # Itens no perfil do usuário
explanation_items_ordered = ['C', 'A', 'E', 'B', 'D']  # itens do perfil do usuário, ordenados por sua importância para explicar a recomendação feita.
relevance = {'A': 3, 'B': 2, 'C': 3, 'D': 1, 'E': 2} # relevâncias dos itens feita pelo user
target_items = ['A', 'B', 'C'] #items recomendados
K_r = 3
K_e = 2

1 - 1: 0 -> ...
```

```python
# Cálculo das métricas
print('POS@K_r, K_e:', pos_at_kr_ke(dummy_rank_fn, user_profile, explanation_items_ordered, K_r, K_e))
print('CDCG@K_e:', cdcg_at_ke(dummy_rank_fn, user_profile, explanation_items_ordered, relevance, cutoff_k=K_r, K_e=K_e))
print('INS@K_e:', ins_at_ke(dummy_score_fn, user_profile, target_items, explanation_items_ordered, K_e))
print('DEL@K_e:', del_at_ke(dummy_score_fn, user_profile, target_items, explanation_items_ordered, K_e))
```
