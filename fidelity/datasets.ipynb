```python

def load_instructrec(path):
    """
    Carrega qualquer dataset .pkl do InstructRec (Books, MovieTV, Yelp, Reads).
    Detecta automaticamente se o conteúdo é list/dict/DataFrame.
    """
    if not os.path.exists(path):
        raise FileNotFoundError(f"Arquivo não encontrado: {path}")

    with open(path, "rb") as f:
        data = pickle.load(f)

    return data


# ============================================
# USO REAL — com o caminho solicitado
# ============================================

path = DATASETS + "/datasets/instructrec/yelpAll_recagent.pkl"
data = load_instructrec(path)

# ADIÇÃO 2: printa tudo SEM truncamento
print(data.iloc[0].keys())
print(data.iloc[0].to_dict())
```

```python
import ast
import pandas as pd

def load_multi_aspect_json(path):
    records = []
    with open(path, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            # Converte "{'a': 1, ...}" em dict Python real
            obj = ast.literal_eval(line)
            records.append(obj)
    return pd.DataFrame(records)
```

```python
beeradvocate_path = "/Users/fillipesantos/Documents/projects/arara_experiments/datasets/multi-aspect-reviews/beeradvocate.json"
ratebeer_path = "/Users/fillipesantos/Documents/projects/arara_experiments/datasets/multi-aspect-reviews/ratebeer.json"

df_beer = load_multi_aspect_json(beeradvocate_path)
df_rate = load_multi_aspect_json(ratebeer_path)

print("BeerAdvocate:", df_beer.shape)
print("RateBeer:", df_rate.shape)

print(df_rate.head())
```
